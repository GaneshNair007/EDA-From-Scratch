Online shopping platforms generate large volumes of transactional data. This data often contains duplicates, incorrect data types, and invalid values such as negative quantities or prices. Cleaning such data is a crucial preprocessing step before performing sales analysis, customer behavior analysis, or predictive modeling.



# **1. Import Libraries**

```python
import pandas as pd
```

# **2. Load the dataset into dataframe**

```python
df=pd.read_excel("/content/Online_Retail.xlsx")
df.head()
```

```python
df.info()
```

Important points about data types to note here:

# **4. Check the duplicates in dataset**

How duplicated() works

1. Compares entire rows

2. If all column values match a previous row → marked True

3. First occurrence = kept, rest = duplicates

How drop_duplicates() works

1. Removes rows where duplicated() == True

2. Keeps only one copy

```python
#Check the count of the duplicate rows
df.duplicated().sum()
```

```python
#Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
#Verify after removing duplicates
df.shape
```

```python
#Verify that all the duplicates are removed
df.duplicated().sum()
```

# **5. Fixing/ Correcting the Data Types**

**Converting InvoiceDate to DateTime**
How it works

1. Converts text → datetime object

2. errors='coerce':

Invalid dates → NaT (Not a Time)

Why needed

Enables:

1. Daily / monthly sales
2. Time-series analysis

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')
```

**Converting CustomerID to Integer**


```python
df['CustomerID'] = df['CustomerID'].astype('Int64')
```

**Fixing Quantity and UnitPrice**
How this works

1. Forces values to numeric

2. Invalid values → NaN

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
```

# **6: Handling Invalid Entries**


**Removing Negative or Zero Quantity**

```python
df = df[df['Quantity'] > 0]
```

**Removing Zero or Negative UnitPrice**

```python
df = df[df['UnitPrice'] > 0]
```

# **7: Handling Missing CustomerID**
How dropna() works

Removes rows where CustomerID is NaN

subset specifies column to check

Why remove instead of filling

  CustomerID cannot be guessed

```python
df = df.dropna(subset=['CustomerID'])
```

# **8: Handling Missing Product Description**

```python
df['Description'] = df['Description'].fillna("Unknown Product")
```

# **9: Final Verification**

```python
df.info()
df.isnull().sum()
```
