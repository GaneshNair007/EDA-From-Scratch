## 5. Statistical Imputation of Missing Values


**1. What is Imputation?**

Imputation is the process of replacing missing values with meaningful estimates so that:

- The dataset becomes complete

- Statistical analysis and ML algorithms can be applied

- Information loss is minimized

**2. Why Imputation is Required**

Most machine learning algorithms:

- Cannot handle missing values

- Produce errors or biased results if NaN values are present

Therefore, missing values must be handled before further analysis.

**3. Types of Statistical Imputation**

Statistical imputation replaces missing values using summary statistics of the data:

- Mean(Average): Numerical (normally distributed)

- Median(Middle value):	Numerical (skewed / outliers)

- Mode(Most frequent value):	Categorical / discrete

To demonstrate imputation techniques, we introduce missing values artificially (for learning purposes).

```python
# Introduce missing values artificially

df_missing = df.copy()
df_missing.loc[5:15, 'chol'] = np.nan
df_missing.loc[0:8, 'trestbps'] = np.nan

df_missing.isnull().sum()
```

```python
# Mean Imputation

mean_imputer  = SimpleImputer(strategy='mean')
df_missing['chol'] = mean_imputer.fit_transform(df_missing[['chol']])

'''OR
df_missing['chol'].fillna(df_missing['chol'].mean(), inplace=True)'''
```

```python
df_missing.isnull().sum()
```

```python
# Median Imputation

median_imputer = SimpleImputer(strategy='median')
df_missing['trestbps'] = median_imputer.fit_transform(df_missing[['trestbps']])

'''OR
df_missing['trestbps'].fillna(df_missing['trestbps'].median(), inplace=True)'''
```

```python
#Mode Imputation (Example)

mode_imputer = SimpleImputer(strategy='most_frequent')
df_missing['Gender'] = mode_imputer.fit_transform(df_missing[['Gender']])

'''OR
df_missing['Gender'].fillna(df_missing['Gender'].mode()[0], inplace=True)'''
```


## 6. k-NN Based Data Imputation


**1. What is k-NN Imputation?**

k-Nearest Neighbors (k-NN) Imputation is an advanced missing-value handling technique that:

- Replaces missing values using similar data points

- Uses the idea that similar patients have similar medical measurements

Unlike mean/median imputation, k-NN preserves relationships between features.

**For a patient record with a missing value:**

- Find k most similar patients (neighbors)

- Measure similarity using distance metrics

- Replace the missing value with:

Mean of neighbors (numerical)

Most frequent value (categorical)

```python
# Introduce missing values artificially

df_knn = df.copy()
df_knn.loc[5:15, 'chol'] = np.nan
df_knn.loc[0:8, 'trestbps'] = np.nan

df_knn.isnull().sum()
```

```python
#Rows with missing values

df_knn[df_knn.isnull().any(axis=1)]
```

```python
knn_imputer = KNNImputer(n_neighbors=5)    #The missing value is filled using 5 most similar records

df_knn = pd.DataFrame(knn_imputer.fit_transform(df), columns=df.columns)
df_knn.head(15)
```
