# Experiment 6: Encoding Categorical Data for ML Readiness

**Use Case:** Customer Churn Prediction

This experiment demonstrates how to prepare categorical data for Machine Learning using:
- Label Encoding
- One-Hot Encoding

The impact of these techniques on model accuracy and performance is also analyzed using google colab.

## Step 1: Import Required Libraries

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

### Explanation
- **pandas / numpy** → data handling
- **LabelEncoder** → converts categories into numbers
- **LogisticRegression** → churn prediction model
- **accuracy_score** → model evaluation

## Step 2: Load Telecom Customer Churn Dataset

```python
df = pd.read_csv("/content/Telco_Customer.csv")
df.head()
```

### Understanding the Data

```python
df.info()
```

### Key Points for Students
- Many columns are **categorical (object type)**
- ML algorithms cannot work directly with text values
- Hence, **encoding is required**

## Step 3: Separate Features and Target

```python
X = df.drop('Churn', axis=1)
y = df['Churn']
```

### Explanation
- **X** → input features
- **y** → target variable (Churn: Yes/No)

## Step 4: LABEL ENCODING

### What is Label Encoding?
- Converts each category into a number
- Example: Yes → 1, No → 0
- Suitable for **binary categorical columns**

```python
df_label = df.copy()
le = LabelEncoder()

for col in df_label.select_dtypes(include='object').columns:
    df_label[col] = le.fit_transform(df_label[col])

df_label.head()
```

## Step 5: Train ML Model using Label Encoded Data

```python
X_label = df_label.drop('Churn', axis=1)
y_label = df_label['Churn']

X_train, X_test, y_train, y_test = train_test_split(
    X_label, y_label, test_size=0.2, random_state=42
)

model_label = LogisticRegression(max_iter=8000)
model_label.fit(X_train, y_train)

y_pred_label = model_label.predict(X_test)
accuracy_label = accuracy_score(y_test, y_pred_label)

accuracy_label
```

## Step 6: ONE-HOT ENCODING

### What is One-Hot Encoding?
- Creates separate columns for each category
- Avoids false ordering
- Commonly used for nominal categorical data

```python
df_onehot = pd.get_dummies(df, drop_first=True)
df_onehot.head()
```

## Step 7: Train ML Model using One-Hot Encoded Data

```python
X_onehot = df_onehot.drop('Churn_Yes', axis=1)
y_onehot = df_onehot['Churn_Yes']

X_train, X_test, y_train, y_test = train_test_split(
    X_onehot, y_onehot, test_size=0.2, random_state=42
)

model_onehot = LogisticRegression(max_iter=8000)
model_onehot.fit(X_train, y_train)

y_pred_onehot = model_onehot.predict(X_test)
accuracy_onehot = accuracy_score(y_test, y_pred_onehot)

accuracy_onehot
```

## Step 8: Comparison of Encoding Techniques

```python
comparison = pd.DataFrame({
    'Encoding Technique': ['Label Encoding', 'One-Hot Encoding'],
    'Accuracy': [accuracy_label, accuracy_onehot]
})

comparison
```

### Final Observations and Key Learnings
- **One-Hot Encoding usually performs better for most Machine Learning models when dealing with categorical data.**
- It avoids incorrect numeric relationships between categories that do not have any natural order.
- Label Encoding is useful only for **binary or ordinal data.**

## Experiment Outcome

- Decide when to use Label vs One-Hot Encoding.
- Students see direct impact of feature engineering on ML model performance.


## 🔹 Feature Scaling (Normalization)

After encoding categorical variables, numerical features may still have different ranges.
Features like `TotalCharges` may have much larger values compared to `tenure` or `MonthlyCharges`.
This can affect the performance of models like **Logistic Regression**.

To solve this issue, **Feature Scaling** is applied.

- **StandardScaler** standardizes data by making the **mean 0** and **standard deviation 1**.  
- **MinMaxScaler** scales the data into a range between **0 and 1**.  

Scaling helps improve model performance and ensures that all features contribute equally.


```python

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python

# Applying StandardScaler
scaler_std = StandardScaler()
X_train_std = scaler_std.fit_transform(X_train)
X_test_std = scaler_std.transform(X_test)

model_std = LogisticRegression(max_iter=1000)
model_std.fit(X_train_std, y_train)

y_pred_std = model_std.predict(X_test_std)
accuracy_std = accuracy_score(y_test, y_pred_std)

accuracy_std
```

```python

# Applying MinMaxScaler
scaler_mm = MinMaxScaler()
X_train_mm = scaler_mm.fit_transform(X_train)
X_test_mm = scaler_mm.transform(X_test)

model_mm = LogisticRegression(max_iter=1000)
model_mm.fit(X_train_mm, y_train)

y_pred_mm = model_mm.predict(X_test_mm)
accuracy_mm = accuracy_score(y_test, y_pred_mm)

accuracy_mm
```

```python

# 🔍 Display Sample Outputs After Scaling

print("---- Sample Values After StandardScaler ----")
print(X_train_std[:5])

print("\nStandardScaler Accuracy:", accuracy_std)

print("\n---- Sample Values After MinMaxScaler ----")
print(X_train_mm[:5])

print("\nMinMaxScaler Accuracy:", accuracy_mm)

# 📊 Final Comparison Table Including Scaling
comparison_scaling = pd.DataFrame({
    'Method': ['Original (One-Hot)', 'StandardScaler', 'MinMaxScaler'],
    'Accuracy': [accuracy_onehot, accuracy_std, accuracy_mm]
})

print("\nFinal Accuracy Comparison:")
comparison_scaling
```


## 🔹 Summary After Encoding and Scaling

In this experiment:

1. Categorical variables were converted into numerical form using **Label Encoding** and **One-Hot Encoding**.
2. Logistic Regression was applied to evaluate model performance.
3. Feature Scaling techniques (**StandardScaler** and **MinMaxScaler**) were applied to normalize numerical features.
4. Model accuracies were compared to observe the impact of encoding and scaling techniques.

This completes the full preprocessing pipeline required for making the dataset ready for Machine Learning.
