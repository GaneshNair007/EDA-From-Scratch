Definition

Data wrangling is the process of cleaning, transforming, structuring, and organizing raw data into a usable format so that it can be easily analyzed and interpreted.

```python
# Import Required Libraries

import pandas as pd
import numpy as np
```

```python
# Load the Dataset

df=pd.read_csv('/content/StudentsPerformance.csv')
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
#Renaming columns for clarity

df = df.rename(columns={
'gender': 'Gender',
'race/ethnicity': 'Ethnicity',
'parental level of education': 'Parental_Education',
'lunch': 'Lunch_Type',
'test preparation course': 'Test_Prep',
'math score': 'Math_Score',
'reading score': 'Reading_Score',
'writing score': 'Writing_Score'
})
```

```python
df.columns
```

```python
# Row Operations

# Students who scored more than 70 in Math
high_math = df[df['Math_Score'] > 70]
high_math
```

```python
#Students who completed test preparation

prep_completed = df[df['Test_Prep'] == 'completed']
prep_completed
```

```python
#Feature Engineering (New Columns)

# Average score across all subjects
df['Average_Score'] = df[['Math_Score', 'Reading_Score', 'Writing_Score']].mean(axis=1)
```

```python
df
```

# Performance Levels Defined
Average Score	Performance Level

0 – 50	Low

50 – 75	Medium

75 – 100	High

```python
#Performance category

df['Performance_Level'] = pd.cut(df['Average_Score'],bins=[0, 50, 75, 100],labels=['Low', 'Medium', 'High'])
df
```

```python
#Grouping Operations

# Average score by gender

avg_by_gender = df.groupby('Gender')[['Math_Score', 'Reading_Score', 'Writing_Score']].mean()
avg_by_gender
```

```python
# Average score by lunch type

avg_by_lunch = df.groupby('Lunch_Type')['Average_Score'].mean()
avg_by_lunch
```

```python
#Sorting and Ranking

#  Sorting students by average score
sorted_df = df.sort_values(by='Average_Score', ascending=False)
sorted_df
```

```python
# Ranking students based on average score

sorted_df['Rank'] = sorted_df['Average_Score'].rank(method='dense', ascending=False)

sorted_df.head()

# method='dense': Students with the same score get the same rank
```

```python
#Merging Multiple Student Datasets (Simulation)

# Creating a second dataset containing attendance information

attendance_df = df[['Gender']].copy()
attendance_df['Attendance_Percentage'] = np.random.randint(60, 100, size=len(attendance_df))
attendance_df
```

```python
# Merging academic performance with attendance data

merged_df = pd.concat([df, attendance_df['Attendance_Percentage']], axis=1)
merged_df
```

```python
# Insights from Merged Dataset

# Relationship between attendance and average score
high_attendance = merged_df[merged_df['Attendance_Percentage'] > 85]
high_attendance
```

```python
merged_df
```

```python
#Final Structured Dataset

final_df = merged_df[['Gender', 'Lunch_Type', 'Test_Prep',
'Math_Score', 'Reading_Score', 'Writing_Score',
'Average_Score', 'Performance_Level', 'Attendance_Percentage']]

final_df
```

```python
final_df.to_csv('structured_student_performance.csv', index=False)
```
