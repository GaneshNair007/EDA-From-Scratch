## Multivariate Exploratory Data Analysis for Smart City Traffic Data

## Aim
To perform multivariate exploratory data analysis using visualization techniques to understand relationships between traffic variables.

## Theory
EDA helps analyze patterns and relationships. Multivariate analysis involves more than two variables.

## Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Explanation:**
- pandas: data handling
- numpy: numerical operations
- matplotlib: basic plotting
- seaborn: advanced visualization

## Load Dataset

```python
df = pd.read_csv("traffic.csv")
df.head()
```

**Explanation:**
- read_csv(): loads dataset
- head(): shows first 5 rows

## Dataset Information

```python
df.info()
```

**Explanation:** Shows data types, columns, and missing values

## Statistical Summary

```python
df.describe()
```

**Explanation:** Gives mean, min, max, standard deviation

## Missing Values

```python
df.isnull().sum()
```

**Explanation:** Counts missing values in each column

## Feature Engineering

```python
df['DateTime'] = pd.to_datetime(df['DateTime'], format="%d/%m/%Y %H:%M:%S")
df['Hour'] = df['DateTime'].dt.hour
```

**Explanation:**
- Converts DateTime to proper format
- Extracts Hour for time-based analysis

## Correlation Matrix

```python
corr = df[['Junction','Vehicles','Hour']].corr()
corr
```

**Explanation:** Shows relationship between variables

## Heatmap

```python
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

**Explanation:** Heatmap visualizes correlation using colors

## Pairplot

```python
sns.pairplot(df[['Junction','Vehicles','Hour']], hue='Junction')
plt.show()
```

**Explanation:**
- Shows relationships between multiple variables
- hue: adds color based on Junction
- Helps compare patterns across categories

## Scatter Plot

```python
sns.scatterplot(x='Junction', y='Vehicles', data=df)
plt.show()
```

**Explanation:** Shows traffic variation across junctions

## Time-based Analysis

```python
sns.lineplot(x='DateTime', y='Vehicles', hue='Junction', data=df)
plt.show()
```

**Explanation:**
- Shows traffic over time
- hue: assigns different colors to each junction
- Helps compare traffic trends

## Hourly Analysis

```python
sns.lineplot(x='Hour', y='Vehicles', hue='Junction', data=df)
plt.show()
```

**Explanation:** Shows traffic pattern at different hours

## Result
Multivariate analysis successfully performed using multiple variables.
